```python
import numpy as np
import scanpy as sc
import anndata as ad
import pandas as pd
import os
import networkx as nx
from matplotlib import pyplot as plt
from matplotlib.lines import Line2D
import sys; sys.path += ['./../../']
from CardamomOT import NetworkModel as CardamomNetworkModel
from CardamomOT.inference.simulations import kon_ref
import matplotlib.gridspec as gridspec
from mpl_toolkits.axes_grid1 import make_axes_locatable
```

```python
def build_gene_subgraph(matrix, gene_names, gene, top_targets=8):
    gene_names = list(gene_names)
    if gene not in gene_names:
        return None, 1.0

    idx = gene_names.index(gene)
    series = pd.Series(matrix[idx, :], index=gene_names).drop(gene, errors='ignore')
    top_idx = series.abs().nlargest(top_targets).index
    max_intensity = float(series.abs().max() or 1.0)

    G = nx.DiGraph()
    for tgt in top_idx:
        w = series[tgt]
        if w != 0:
            G.add_edge(gene, tgt, weight=float(w))
    return G, max_intensity


def draw_gene_subgraph(ax, G, gene, max_intensity):
    if G is None or G.number_of_edges() == 0:
        ax.text(0.5, 0.5, f"{gene}\n(no GRN data)", ha='center', va='center',
                transform=ax.transAxes, fontsize=6, color='gray')
        ax.axis('off')
        return

    pos = nx.spring_layout(G, seed=42, pos={gene: (0.0, 0.0)}, fixed=[gene])
    node_colors = ['#4C9BE8' if n == gene else '#E8E8E8' for n in G.nodes]
    node_sizes  = [800 if n == gene else 600 for n in G.nodes]
    nx.draw_networkx_nodes(G, pos, node_color=node_colors, node_size=node_sizes, ax=ax)
    nx.draw_networkx_labels(G, pos, font_size=5.5, ax=ax)
    ax.margins(0.15)

    edges_pos = [(u, v, d['weight']) for u, v, d in G.edges(data=True) if d['weight'] > 0]
    edges_neg = [(u, v, d['weight']) for u, v, d in G.edges(data=True) if d['weight'] < 0]
    width_scale = 3.0 / max_intensity
    kw = dict(ax=ax, connectionstyle='arc3,rad=0.1', min_target_margin=15)

    if edges_pos:
        nx.draw_networkx_edges(G, pos,
                               edgelist=[(u, v) for u, v, _ in edges_pos],
                               edge_color='#2ECC71',
                               width=[abs(w) * width_scale for *_, w in edges_pos],
                               arrows=True, arrowsize=8, **kw)
    if edges_neg:
        nx.draw_networkx_edges(G, pos,
                               edgelist=[(u, v) for u, v, _ in edges_neg],
                               edge_color='#E74C3C',
                               width=[abs(w) * width_scale for *_, w in edges_neg],
                               arrows=True,
                               arrowstyle='-[,widthB=0.8,lengthB=0.0',
                               **kw)

    ax.set_title(f"Targets of {gene}", fontsize=6.5, fontweight='bold', pad=2)
    ax.axis('off')


def draw_full_grn(
    ax,
    matrix,
    gene_names,
    priority_genes=None,
    priority_edges_per_gene=4,
    extra_top_edges=30,
    gene_pool_size=10,
    favor_high_power=True,
    high_power_quantile=0.75,
    low_target_quantile=0.25,
    low_target_incoming_cap=2,
):
    gene_names = list(gene_names)
    n = len(gene_names)
    gene_idx = {g: i for i, g in enumerate(gene_names)}

    all_abs = np.abs(matrix)
    max_w = all_abs.max() or 1.0
    reg_power = all_abs.sum(axis=1)

    high_thr = float(np.quantile(reg_power, high_power_quantile)) if reg_power.size else 0.0
    low_thr = float(np.quantile(reg_power, low_target_quantile)) if reg_power.size else 0.0

    is_high_power = {g: reg_power[gene_idx[g]] >= high_thr for g in gene_names}
    is_low_power_target = {g: reg_power[gene_idx[g]] <= low_thr for g in gene_names}

    G = nx.DiGraph()
    G.add_nodes_from(gene_names)

    off_diag = ~np.eye(n, dtype=bool)
    rows, cols = np.where(off_diag)

    selected_edges = set()
    incoming_counts = {g: 0 for g in gene_names}
    outgoing_counts = {g: 0 for g in gene_names}
    effective_priority_genes = []

    def can_add_target(tgt_name):
        if is_low_power_target[tgt_name] and incoming_counts[tgt_name] >= int(low_target_incoming_cap):
            return False
        return True

    def try_add_edge(src_name, tgt_name):
        if src_name == tgt_name:
            return False
        edge = (src_name, tgt_name)
        if edge in selected_edges:
            return False
        if not can_add_target(tgt_name):
            return False
        selected_edges.add(edge)
        incoming_counts[tgt_name] += 1
        outgoing_counts[src_name] += 1
        return True

    if priority_genes is not None:
        effective_priority_genes = [g for g in priority_genes if g in gene_idx][:int(gene_pool_size)]

        for src_name in effective_priority_genes:
            src_idx = gene_idx[src_name]
            row_abs = all_abs[src_idx, :].copy()
            row_abs[src_idx] = 0.0

            valid_tgt_idx = np.where(row_abs > 0)[0]
            if valid_tgt_idx.size == 0:
                continue

            k = min(int(priority_edges_per_gene), valid_tgt_idx.size)
            ranked_tgt = valid_tgt_idx[np.argsort(row_abs[valid_tgt_idx])[::-1][:k]]
            for tgt_idx in ranked_tgt:
                try_add_edge(src_name, gene_names[tgt_idx])

    candidate_edges = []
    for r, c in zip(rows, cols):
        w = float(matrix[r, c])
        if w == 0:
            continue

        src_name = gene_names[r]
        tgt_name = gene_names[c]
        base = abs(w)

        if favor_high_power:
            both_high = is_high_power[src_name] and is_high_power[tgt_name]
            one_high = is_high_power[src_name] or is_high_power[tgt_name]
            if both_high:
                base *= 3
            elif one_high:
                base *= 2

        candidate_edges.append((src_name, tgt_name, base))

    target_extra_edges = max(0, int(extra_top_edges))
    added_extra = 0

    while added_extra < target_extra_edges and candidate_edges:
        best_i = -1
        best_score = -np.inf

        for i, (src_name, tgt_name, base_score) in enumerate(candidate_edges):
            if (src_name, tgt_name) in selected_edges:
                continue
            if not can_add_target(tgt_name):
                continue

            adjusted_score = base_score / (1.0 + 1 * incoming_counts[tgt_name]
                                           + 1 * outgoing_counts[src_name])
            if adjusted_score > best_score:
                best_score = adjusted_score
                best_i = i

        if best_i < 0:
            break

        src_name, tgt_name, _ = candidate_edges[best_i]
        if try_add_edge(src_name, tgt_name):
            added_extra += 1

    for src_name, tgt_name in selected_edges:
        G.add_edge(
            src_name,
            tgt_name,
            weight=float(matrix[gene_idx[src_name], gene_idx[tgt_name]]),
        )

    if G.number_of_edges() == 0:
        ax.text(0.5, 0.5, 'No GRN edges after filtering', ha='center', va='center',
                transform=ax.transAxes, fontsize=6, color='gray')
        ax.axis('off')
        return

    # Spring layout on undirected graph with edge weights
    G_und = G.to_undirected()
    for u, v in G_und.edges():
        ws = [abs(G[a][b]['weight']) for a, b in ((u, v), (v, u)) if G.has_edge(a, b)]
        G_und[u][v]['weight'] = float(np.mean(ws))

    # Connectivity activity (number of in + out edges) to bias center placement
    activity = {
        nd: float(G.in_degree(nd) + G.out_degree(nd))
        for nd in G.nodes
    }
    max_activity = max(activity.values()) if activity else 1.0

    # Initialize connected nodes near center, weakly connected/isolated farther away
    pos_init = {}
    if n > 1:
        nodes_sorted = sorted(G.nodes, key=lambda nd: activity[nd], reverse=True)
        for rank, nd in enumerate(nodes_sorted):
            a_norm = activity[nd] / (max_activity or 1.0)
            radius = 0.08 + (1.0 - a_norm) * 0.92
            angle = 2.0 * np.pi * (rank / max(1, n))
            jitter = 0.04 * (1.0 - a_norm)
            pos_init[nd] = np.array([
                radius * np.cos(angle) + jitter * np.cos(3.0 * angle),
                radius * np.sin(angle) + jitter * np.sin(5.0 * angle),
            ])

    pos = nx.spring_layout(
        G_und,
        weight='weight',
        seed=42,
        k=6.0 / n**0.5,
        iterations=200,
        pos=pos_init,
    )

    # All-pairs force pass: short-range repulsion + long-range attraction
    nodes_list = list(pos)
    if len(nodes_list) > 1:
        ideal_dist = 0.14
        repulsion_strength = 0.025
        attraction_strength = 0.010
        center_pull_strength = 0.012
        step_size = 0.15

        activity_arr = np.array([activity[nd] for nd in nodes_list], dtype=float)
        activity_arr = activity_arr / (activity_arr.max() or 1.0)

        for _ in range(120):
            pos_arr = np.array([pos[nd] for nd in nodes_list])
            diff = pos_arr[:, None, :] - pos_arr[None, :, :]          # (N, N, 2)
            dist = np.linalg.norm(diff, axis=-1)                      # (N, N)

            pair_mask = ~np.eye(len(nodes_list), dtype=bool)
            safe_dist = np.where(pair_mask, dist, 1.0)

            unit = np.zeros_like(diff)
            nonzero = pair_mask & (safe_dist > 1e-9)
            unit[nonzero] = diff[nonzero] / safe_dist[nonzero][:, None]

            repel_mag = repulsion_strength * np.clip(ideal_dist - safe_dist, 0.0, None) / ideal_dist
            attract_mag = attraction_strength * np.clip(safe_dist - ideal_dist, 0.0, None) / ideal_dist
            net_mag = (repel_mag - attract_mag) * pair_mask

            delta = (unit * net_mag[:, :, None]).sum(axis=1)

            # Weak radial pull of high-activity nodes toward center
            delta += (-pos_arr) * (center_pull_strength * activity_arr)[:, None]

            max_disp = np.linalg.norm(delta, axis=1).max()
            if max_disp < 1e-4:
                break

            for i, nd in enumerate(nodes_list):
                pos[nd] += step_size * delta[i]

    # Node appearance: colour by regulatory power
    node_power = np.array([reg_power[gene_idx[nd]] for nd in G.nodes]) / (reg_power.max() or 1.0)
    cmap = plt.cm.get_cmap('YlOrRd')
    nx.draw_networkx_nodes(G, pos, node_color=[cmap(p) for p in node_power],
                           node_size=100, alpha=1, ax=ax)
    nx.draw_networkx_labels(G, pos, font_size=3, ax=ax)
    ax.margins(0.12)

    # Edges
    edges_pos = [(u, v) for u, v, d in G.edges(data=True) if d['weight'] > 0]
    edges_neg = [(u, v) for u, v, d in G.edges(data=True) if d['weight'] < 0]
    widths = lambda el: [abs(G[u][v]['weight']) / max_w * 2.5 for u, v in el]
    kw = dict(arrows=True, arrowsize=7, connectionstyle='arc3,rad=0.12', min_target_margin=5, ax=ax)
    if edges_pos:
        nx.draw_networkx_edges(G, pos, edgelist=edges_pos,
                               edge_color='#2ECC71', width=widths(edges_pos), alpha=0.6, **kw)
    if edges_neg:
        nx.draw_networkx_edges(G, pos, edgelist=edges_neg,
                               edge_color='#E74C3C', width=widths(edges_neg), alpha=0.6, **kw)

    ax.legend(handles=[
        Line2D([0], [0], color='#2ECC71', linewidth=1.5, label='Activation'),
        Line2D([0], [0], color='#E74C3C', linewidth=1.5, label='Inhibition'),
    ], fontsize=5, loc='lower right', framealpha=0.7, borderpad=0.4)

    ax.set_title(
        f"GRN ({len(selected_edges)} edges, high regulatory power biased)",
        fontsize=6.5, fontweight='bold', pad=2
    )
    ax.axis('off')
```

```python
def load_grn_matrix(p):
    full_path = os.path.join(p, 'Data', 'data_full.h5ad')
    train_path = os.path.join(p, 'Data', 'data_train.h5ad')
    adata = sc.read_h5ad(full_path if os.path.exists(full_path) else train_path)
    genes = list(adata.var_names)

    inter = np.load(os.path.join(p, 'cardamom', 'inter_simul.npy'))
    basal = np.load(os.path.join(p, 'cardamom', 'basal_simul.npy'))
    mixture_parameters = np.load(os.path.join(p, 'cardamom', 'mixture_parameters.npy'))

    ks_cells = np.max(mixture_parameters[:-1], axis=0)
    ks = (mixture_parameters[:-1] / ks_cells).T

    P = np.load(os.path.join(p, 'cardamom', 'data_prot.npy'))  # (n_cells, n_genes)

    def compute_kon_for_P(P_input):
        kon_vals = []
        for cell in range(P_input.shape[0]):
            kon_cell = kon_ref(P_input[cell].reshape(1, -1), ks, inter, basal)
            kon_vals.append(kon_cell)
        return np.array(kon_vals)  # (n_cells, n_genes)

    kon_vector = compute_kon_for_P(P)

    grn_mat = inter[1:, 1:]
    matrix = grn_mat[:, :, 0] if grn_mat.ndim == 3 else grn_mat

    n_genes = P.shape[1]
    kon_zero_by_gene = np.zeros((n_genes, P.shape[0], n_genes), dtype=kon_vector.dtype)
    regulatory_power = np.zeros(n_genes, dtype=np.float32)

    for g in range(n_genes):
        P_gene_zero = P.copy()
        P_gene_zero[:, g] = 0.0
        kon_gene_zero = compute_kon_for_P(P_gene_zero)
        kon_zero_by_gene[g] = kon_gene_zero
        regulatory_power[g] = np.sum(np.abs(kon_vector - kon_gene_zero))

    return matrix, genes, kon_vector, P, regulatory_power[1:]


def signed_log1p(x):
    """Transforme de façon symétrique: sign(x) * log1p(|x|)."""
    return np.sign(x) * np.log1p(np.abs(x))


def figure_sup1_3(
    datapath,
    figname='figure_sup1_3.pdf',
    top_k=100,
    min_label_gap='auto',
    auto_gap_quantile=0.08,
    label_angle=12,
):
    matrix, gene_names, _, _, regulatory_power = load_grn_matrix(datapath)
    reg_power_raw = np.asarray(regulatory_power)

    # Top 10 genes by regulatory power (used for both panel A and panel C)
    top10_idx = np.argsort(reg_power_raw)[::-1][:10]
    top10_genes = [gene_names[i] for i in top10_idx]

    fig = plt.figure(figsize=(8.27, 11.69))
    gs   = gridspec.GridSpec(2, 1, figure=fig, height_ratios=[2, 1],
                            hspace=0.1, wspace=0.28)
    gs01 = gs[0, 0].subgridspec(2, 1, hspace=0.1, wspace=0.2)
    gs02 = gs[1, 0].subgridspec(2, 2, hspace=0.1, wspace=0.2)
    axes = (
    [fig.add_subplot(gs01[r, c]) for r in range(2) for c in range(1)] +
    [fig.add_subplot(gs02[r, c]) for r in range(2) for c in range(2)] )
    for ax, label, pos in zip(axes[:2], "AB", [-0.04, -0.35]):
        ax.text(pos, 1.05, label, transform=ax.transAxes,
                ha="left", va="bottom", fontsize=8, fontweight="bold", clip_on=False)

    # Full GRN network in 1st plot of gs01: 4 strongest edges for top 10 genes + 30 extra strongest edges
    draw_full_grn(
        axes[0],
        matrix,
        gene_names,
        priority_genes=top10_genes,
        priority_edges_per_gene=4,
        extra_top_edges=top_k,
        gene_pool_size=10,
    )

    # Heatmap of full GRN in 2nd plot of gs01 (signed log transform)
    matrix_heatmap = signed_log1p(matrix)
    extremum = np.max(np.abs(matrix_heatmap))
    im = axes[1].imshow(matrix_heatmap, aspect='equal', cmap='RdYlGn', vmin=-extremum, vmax=extremum)
    axes[1].set_box_aspect(1)
    divider = make_axes_locatable(axes[1])
    cax = divider.append_axes("right", size="4%", pad=0.05)
    fig.colorbar(im, cax=cax)

    # Violin plot of gene regulatory power to the right of the heatmap
    reg_power = np.log1p(reg_power_raw)
    reg_span = float(np.ptp(reg_power)) if reg_power.size else 0.0
    if isinstance(min_label_gap, str) and min_label_gap.lower() == 'auto':
        effective_label_gap = max(0.02, auto_gap_quantile * reg_span)
    else:
        effective_label_gap = float(min_label_gap)
    vax = divider.append_axes("right", size="50%", pad=0.80)
    vax.text(-0.35, 1.05, 'C', transform=vax.transAxes,
             ha="left", va="bottom", fontsize=8, fontweight="bold", clip_on=False)
    vax.violinplot(reg_power, showmeans=True, showextrema=False)

    # Overlay top 10 genes as points + labels
    top10_power = reg_power[top10_idx]
    x_top10 = np.ones(len(top10_power))
    vax.scatter(x_top10, top10_power, s=4, color='black', alpha=0.85, zorder=3)

    right_labeled_y = []
    left_labeled_y = []
    for y_val, gene in zip(top10_power, top10_genes):
        can_place_right = all(abs(y_val - prev_y) >= effective_label_gap for prev_y in right_labeled_y)

        if can_place_right:
            x_txt, ha, rot = 1.06, 'left', label_angle
            right_labeled_y.append(y_val)
        else:
            can_place_left = all(abs(y_val - prev_y) >= effective_label_gap for prev_y in left_labeled_y)
            if not can_place_left:
                continue
            x_txt, ha, rot = 0.94, 'right', -label_angle
            left_labeled_y.append(y_val)

        vax.text(
            x_txt,
            y_val,
            gene,
            fontsize=5.5,
            va='center',
            ha=ha,
            rotation=rot,
            rotation_mode='anchor',
            clip_on=False,
        )

    vax.tick_params(axis='y', labelsize=6)
    vax.set_title('Gene regulatory power', fontsize=6.5, fontweight='bold', pad=2)

    axes[1].set_title('GRN (signed log1p)')

    # Small GRNs for top 4 genes by regulatory power
    top4_idx = np.argsort(reg_power)[::-1][:4]
    top4_genes = [gene_names[i] for i in top4_idx]

    for ax, label in zip(axes[2:], "DEFG"):
        ax.text(-0.08, 1.05, label, transform=ax.transAxes,
                ha="left", va="bottom", fontsize=8, fontweight="bold", clip_on=False)

    for ax, gene in zip(axes[2:], top4_genes):
        G, max_intensity = build_gene_subgraph(matrix, gene_names, gene)
        draw_gene_subgraph(ax, G, gene, max_intensity)

    plt.savefig(figname, dpi=600, bbox_inches='tight')
    plt.show()
```
